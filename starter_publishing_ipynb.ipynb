```
import re
import string

def preprocess_text(text):
    # 1. 본문 추출 (Project Gutenberg 시작과 끝 마커 사이의 내용만 추출)
    # re.DOTALL을 사용하여 줄바꿈(\n)도 마침표(.)에 포함되도록 합니다.
    body_pattern = r"\*\*\* START OF THE PROJECT GUTENBERG EBOOK.*?\*\*\*(.*?)\*\*\* END OF THE PROJECT GUTENBERG EBOOK"
    match = re.search(body_pattern, text, flags=re.DOTALL | re.IGNORECASE)

    if match:
        text = match.group(1)

    # 2. 제목, 작가, 목차(Contents) 제거
    # 'Contents' 단어부터 첫 번째 장(I. 또는 CHAPTER I 등)이 시작되기 전까지를 찾아 삭제합니다.
    # 보통 목차는 'Contents'로 시작하여 챕터 리스트가 나오고, 본문은 다시 'I.' 등으로 시작합니다.
    toc_pattern = r"(?is).*?contents\s+.*?(?=\n\s*I\.|\n\s*CHAPTER I|\n\s*ADVENTURE I)"
    text = re.sub(toc_pattern, "", text, count=1)

    # --- 기존 로직 유지 ---
    text = text.replace("Project Gutenberg", "")
    text = text.replace("Gutenberg", "")

    # Remove carriage returns
    text = text.replace("\r", "")

    # fix quotes
    text = text.replace("“", "\"")
    text = text.replace("”", "\"")

    # 대소문자 마커 처리 (^ 기호 삽입)
    # 단어 시작의 대문자 처리
    text = re.sub(r"(?<![a-zA-Z])([A-Z])", lambda match: f"^{match.group(0).lower()}", text)
    # 그 외 모든 대문자를 소문자로 변경
    text = re.sub(r"([A-Z])", lambda match: f"{match.group(0).lower()}", text)

    # 공백 및 특수 토큰 처리 (순서가 중요합니다)
    # 탭과 줄바꿈을 먼저 처리한 후 일반 공백을 처리합니다.
    text = text.replace("\t", " zztabzz ")
    text = text.replace("\n", " zznewlinezz ")

    # 연속된 공백 하나로 압축 및 특수 토큰화
    text = re.sub(r" +", " ", text)
    text = text.replace(" ", " zzspacezz ")

    # 구두점 분리 (기존 로직)
    for punctuation in string.punctuation:
        if punctuation == "^": # 우리가 사용중인 마커는 제외
            continue
        text = text.replace(punctuation, f" {punctuation} ")

    return text
```

```
def postprocess_text(text):

    # Replace special words with whitespace characters
    text = text.replace("zztabzz", "\t")
    text = text.replace("zznewlinezz", "\n")
    text = text.replace("zzspacezz", " ")

    # Remake capital letters at beginning of words
    text = re.sub(r"\^([a-z])", lambda match: f"{match.group(1).upper()}", text)

    text = text.replace("^", "")

    return text
```

```
def getMyText():
    # 1. URL 리스트 정의 (기존 코드에서 정의한 부분)
    gutenberg_prefix = 'https://www.gutenberg.org/ebooks/'
    num_list = [1661, 834, 2350, 69700, 108, 2097, 3070, 3289, 244]
    gutenberg_suffix = '.txt.utf-8'
    url_list = [f"{gutenberg_prefix}{num}{gutenberg_suffix}" for num in num_list]

    local_dir = 'saved_files'
    if not os.path.exists(local_dir):
        os.makedirs(local_dir)

    full_text = "" # 모든 책의 내용을 담을 변수

    for i, file_url in enumerate(url_list):
        # 각 파일마다 고유한 이름을 부여 (예: book_0.txt, book_1.txt...)
        file_name = f'book_{i}.txt'
        local_path = os.path.join(local_dir, file_name)

        if os.path.exists(local_path):
            print(f"File '{file_name}' found locally. Using it.")
        else:
            print(f"Downloading '{file_name}' from {file_url}...")
            # 리스트의 개별 URL을 하나씩 다운로드
            downloaded_path = tf.keras.utils.get_file(file_name, file_url)

            # 다운로드된 파일을 지정된 디렉토리로 복사
            with open(downloaded_path, 'rb') as source_file:
                with open(local_path, 'wb') as dest_file:
                    dest_file.write(source_file.read())

        # 저장된 파일을 읽어서 전체 텍스트 변수에 누적
        with open(local_path, 'rb') as f:
            full_text += f.read().decode(encoding='utf-8') + "\n\n"

    # 모든 텍스트가 합쳐진 full_text를 전처리함수에 전달
    return preprocess_text(full_text)
```

```
# This function will generate our sequence pairs:
def split_input_target(sequence):
    input_ids = sequence[:-1]
    target_ids = sequence[1:]
    return input_ids, target_ids

# This function will create the dataset
def text_to_dataset(text):
  all_ids = vectorize_layer(text)
  ids_dataset = tf.data.Dataset.from_tensor_slices(all_ids)
  del all_ids
  sequences = ids_dataset.batch(sequence_length+1, drop_remainder=True)
  del ids_dataset

  # Call the function for every sequence in our list to create a new dataset
  # of input->target pairs
  dataset = sequences.map(split_input_target)
  del sequences

  # shuffle


  return dataset
```

## III. Build the model

Next, we'll build our model. Up until this point, you've been using the Keras symbolic, or imperative API for creating your models. Doing something like:

    model = tf.keras.models.Sequentla()
    model.add(tf.keras.layers.Dense(80, activation='relu))
    etc...

However, tensorflow has another way to build models called the Functional API, which gives us a lot more control over what happens inside the model. You can read more about [the differences and when to use each here](https://blog.tensorflow.org/2019/01/what-are-symbolic-and-imperative-apis.html).

We'll use the functional API for our RNN in this example. This will involve defining our model as a custom subclass of `tf.keras.Model`.

If you're not familiar with classes in python, you might want to review [this quick tutorial](https://www.w3schools.com/python/python_classes.asp), as well as [this one on class inheritance](https://www.w3schools.com/python/python_inheritance.asp).

Using a functional model is important for our situation because we're not just training it to predict a single character for a single sequence, but as we make predictions with it, we need it to remember those predictions as use that memory as it makes new predictions.

```
class AustenTextModel(tf.keras.Model):
    def __init__(self, vocab_size, embedding_dim, rnn_units):
        super().__init__()
        # 1. Embedding Layer
        self.embedding = tf.keras.layers.Embedding(vocab_size, embedding_dim)

        # 2. Reduced LSTM Layers (2-stack)
        self.lstm1 = tf.keras.layers.LSTM(rnn_units, return_sequences=True, return_state=True, dropout=0.2)
        self.lstm2 = tf.keras.layers.LSTM(rnn_units, return_sequences=True, return_state=True, dropout=0.2)

        # 3. Dense & Dropout Layers
        self.hidden1 = tf.keras.layers.Dense(rnn_units, activation='relu')
        self.dropout = tf.keras.layers.Dropout(0.3)
        self.dense = tf.keras.layers.Dense(vocab_size)

    def call(self, inputs, states=None, return_state=False, training=False):
        x = inputs
        x = self.embedding(x, training=training)

        batch_size = tf.shape(inputs)[0]

        # 2개 레이어에 맞게 초기 상태 설정
        if states is None:
            # LSTM의 state는 [h, c] 리스트 형태임
            states1 = [tf.zeros([batch_size, self.lstm1.units]), tf.zeros([batch_size, self.lstm1.units])]
            states2 = [tf.zeros([batch_size, self.lstm2.units]), tf.zeros([batch_size, self.lstm2.units])]
        else:
            states1 = states[0]
            states2 = states[1]

        # LSTM 순방향 연산
        x, state_h1, state_c1 = self.lstm1(x, initial_state=states1, training=training)
        x, state_h2, state_c2 = self.lstm2(x, initial_state=states2, training=training)

        # 출력용 상태 리스트 구성
        states_out = [[state_h1, state_c1], [state_h2, state_c2]]

        # Dense 레이어 및 규제 적용
        x = self.hidden1(x, training=training)
        x = self.dropout(x, training=training)
        x = self.dense(x, training=training)

        if return_state:
            return x, states_out
        else:
            return x
```

```
# Here's the code we'll use to sample for us. It has some extra steps to apply
# the temperature to the distribution, and to make sure we don't get empty
# characters in our text. Most importantly, it will keep track of our model
# state for us.

class OneStep(tf.keras.Model):
  def __init__(self, model, vectorize_layer, vocabulary, temperature=1):
    super().__init__()
    self.temperature=temperature
    self.model = model
    self.vectorize_layer = vectorize_layer
    self.vocabulary = vocabulary
    #print("initialized")

    # Create a mask to prevent "" or "[UNK]" from being generated.
    skip_ids = StringLookup(vocabulary=list(vocabulary))(['', '[UNK]'])[:, None]
    #print(skip_ids)
    #print("3")
    sparse_mask = tf.SparseTensor(
        # Put a -inf at each bad index.
        values=[-float('inf')]*len(skip_ids),
        indices = skip_ids,
        # Match the shape to the vocabulary
        dense_shape=[len(vocabulary)])
    #print("4")
    self.prediction_mask = tf.sparse.to_dense(sparse_mask,validate_indices=False)
    #print("5")

  @tf.function
  def generate_one_step(self, inputs, states=None):
    # Convert strings to token IDs.
    #input_chars = tf.strings.unicode_split(inputs, 'UTF-8')
    input_ids = self.vectorize_layer(inputs)
    #print(input_ids)

    # Run the model.
    # predicted_logits.shape is [batch, char, next_char_logits]
    predicted_logits, states =  self.model(inputs=input_ids, states=states,
                                          return_state=True)
    del input_ids
    # Only use the last prediction.
    predicted_logits = predicted_logits[:, -1, :]
    predicted_logits = predicted_logits/self.temperature

    # Apply the prediction mask: prevent "" or "[UNK]" from being generated.
    predicted_logits = predicted_logits + self.prediction_mask

    # Sample the output logits to generate token IDs.
    predicted_ids = tf.random.categorical(predicted_logits, num_samples=1)
    del predicted_logits
    predicted_ids = tf.squeeze(predicted_ids, axis=-1)

    #print(predicted_ids[0])

    # Return the characters and model state.
    return words_from_ids(predicted_ids), states
```

```
def produce_sample(model, vectorize_layer, vocabulary, temp, epoch, prompt):
  # Create an instance of the character generator
  #print("entered")
  one_step_model = OneStep(model, vectorize_layer, vocabulary, temp)
  #print("rand one step")
  # Now, let's generate a 1000 character chapter by giving our model "Chapter 1"
  # as its starting text
  states = None
  next_char = tf.constant([preprocess_text(prompt)])
  result = [tf.constant([prompt])]

  for n in range(200):
    next_char, states = one_step_model.generate_one_step(next_char, states=states)
    #print(next_char)
    result.append(next_char)
    #print(result)

  result = tf.strings.join(result)
  #print(result)

  # Print the results formatted.
  #print('Temp: ' + str(temp) + '\n')
  print(postprocess_text(result[0].numpy().decode('utf-8')))
  #print('\n\n')
  print('Epoch: ' + str(epoch) + '\n', file=open(path + 'tree.txt', 'a'))
  print('Temp: ' + str(temp) + '\n', file=open(path + 'tree.txt', 'a'))
  print(postprocess_text(result[0].numpy().decode('utf-8')), file=open(path + 'tree.txt', 'a'))
  print('\n\n', file=open(path + 'tree.txt', 'a'))
  del states
  del next_char
  del result
```
